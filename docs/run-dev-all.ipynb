# Run /dev/all from source

A runbook for this repository. Open it in /dev/all (add the repo as a project, or open the file in Jupyter /
VS Code — it's a standard notebook) and run the cells top to bottom. Cells run from the repo root.
Prerequisites: Rust 1.91+, Node 22, and the [Tauri prerequisites](https://v2.tauri.app/start/prerequisites/)
(WebView2 is preinstalled on Windows 11; Xcode Command Line Tools on macOS).

## 1. Check the toolchain

```shell
rustc --version && cargo --version
node --version && npm --version
git --version
```

## 2. Install the desktop app's dependencies

```shell
cd apps/desktop && npm install
```

## 3. Run the desktop app (dev mode)

Starts Vite and the Tauri shell with hot reload. The cell keeps running while the app is open — press ■ to stop
it (that ends the whole process tree).

```shell
cd apps/desktop && npm run app
```

On Windows, to drive the real window over the Chrome DevTools protocol (see `docs/IMPLEMENTATION_PLAN.md` §5),
start it with remote debugging instead:

```shell
$env:WEBVIEW2_ADDITIONAL_BROWSER_ARGUMENTS = '--remote-debugging-port=9223'
Set-Location apps/desktop
npm run app
```

## 4. Tests and checks

```shell
cargo test --workspace
```

```shell
cargo clippy --workspace --all-targets -- -D warnings
```

```shell
cd apps/desktop && npx tsc --noEmit -p tsconfig.json && npm test
```

## 5. The CLI against the demo project

```shell
cargo run -q -p devall-cli -- run examples/jsonplaceholder --env public
```
